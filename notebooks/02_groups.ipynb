# 02 Trend groups

The logic lives in `pipeline/tract/groups.py`.

- Six features: yearly growth of jobs, people and real rent, before and after (see notebook 01)
- Clip each to its 1st and 99th percentile, then standardize
- k from 2 to 10: elbow, silhouette, Davies-Bouldin, gap statistic, stability across seeds and subsamples
- K-means vs a Gaussian mixture vs Ward clustering
- Rule for k: stable (subsample adjusted Rand at least 0.8), no group under 2% of tracts, then the best silhouette

**Output**: `trend_group` column in `trends_tract.csv`, `processed/5500/groups_*`

The same k checks, interactive. Pick a score in the menu. Hover a point for its value. The chosen k is the dark point.

```python
import altair as alt
import geopandas as gpd

alt.renderers.enable("html")            # the chart works in the saved HTML page too
alt.data_transformers.disable_max_rows()

import json
from pipeline.tract.io import out_dir

# Read the saved results of groups.run() above, so this cell runs on its own
km = pd.read_csv(out_dir() / "groups_k_metrics.csv")
K = json.loads((out_dir() / "groups_summary.json").read_text())["k"]
scores = {"silhouette": "Silhouette (higher is better)", "davies_bouldin": "Davies-Bouldin (lower is better)",
          "gap": "Gap statistic (higher is better)", "ari_seeds": "Stability across seeds, ARI",
          "ari_bootstrap": "Stability across subsamples, ARI", "smallest_group_share": "Smallest group share"}
long = km.melt(id_vars="k", value_vars=[c for c in scores if c in km.columns], var_name="score", value_name="value")
long["score"] = long.score.map(scores)
long["chosen"] = long.k == K

pick = alt.param(name="score", value=scores["silhouette"],
                 bind=alt.binding_select(options=list(scores.values()), name="Score "))
base = alt.Chart(long).transform_filter(alt.datum.score == pick)
line = base.mark_line(color="#7fb2ad").encode(x=alt.X("k:O", title="k (groups)"), y=alt.Y("value:Q", title=None, scale=alt.Scale(zero=False)))
dots = base.mark_circle(size=90).encode(
    x="k:O", y="value:Q",
    color=alt.condition(alt.datum.chosen, alt.value("#143c3c"), alt.value("#aed0cc")),
    tooltip=["k:O", alt.Tooltip("value:Q", format=".3f")],
)
(line + dots).add_params(pick).properties(width=420, height=260, title="K-means scores by k")
```

### Explore the groups

Left: each tract by its growth before and after 2019, coloured by group. Click a group in the legend to pick it (shift-click for more). Right: the map you can zoom and hover. The groups are descriptive, not natural clusters (silhouette 0.136).

```python
tg = pd.read_csv(ROOT / "data/processed/trends_tract.csv", dtype={"GEOID": str})
tg = tg[tg.trend_group.notna() & (tg.trend_group != "Not enough data")]
pts = tg[["GEOID", "trend_group", "growth_before", "growth_after"]].dropna().copy()
pts[["growth_before", "growth_after"]] = (pts[["growth_before", "growth_after"]] * 100).clip(-6, 6).round(2)
order = list(pts.trend_group.value_counts().index)
palette = ["#94a3b8", "#60a5fa", "#f59e0b", "#4f918b", "#f472b6", "#34d399", "#ef4444", "#14b8a6"][: len(order)]

grp = alt.selection_point(fields=["trend_group"], bind="legend")
alt.Chart(pts).mark_circle(size=16).encode(
    x=alt.X("growth_before:Q", title="Growth before 2019 (% a year, combined)"),
    y=alt.Y("growth_after:Q", title="Growth after 2019 (% a year, combined)"),
    color=alt.Color("trend_group:N", title="Trend group", scale=alt.Scale(domain=order, range=palette)),
    opacity=alt.condition(grp, alt.value(0.8), alt.value(0.05)),
    tooltip=[alt.Tooltip("GEOID:N", title="Tract"), alt.Tooltip("trend_group:N", title="Group"), "growth_before:Q", "growth_after:Q"],
).add_params(grp).properties(width=460, height=340, title="Tracts by trend group")
```

```python
import shapely
import warnings

warnings.simplefilter("ignore", RuntimeWarning)   # shapely warns on a few tract rings while simplifying. The shapes are fine
g = gpd.read_file(ROOT / "data/processed/tracts_2020.gpkg")[["GEOID", "NAMELSAD", "geometry"]].to_crs(4326)
g["geometry"] = shapely.set_precision(g.geometry.make_valid().simplify(0.004, preserve_topology=True).values, 0.001)  # about 400 m: a small page
g = g.merge(tg[["GEOID", "trend_group"]], on="GEOID")
g.explore(
    column="trend_group", categories=order, cmap=palette, tiles="Esri.WorldGrayCanvas",
    tooltip=["NAMELSAD", "trend_group"], tooltip_kwds={"aliases": ["Tract", "Group"]},
    style_kwds={"weight": 0.2, "color": "white", "fillOpacity": 0.75},
    legend_kwds={"caption": "Trend group"},
)
```
